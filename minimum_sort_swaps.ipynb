```python
input_list = eval(input())

swaps = 0
i = 0

while i < len(input_list):
    correct_pos = input_list[i] - 1
    
    if input_list[i] != input_list[correct_pos]:
        input_list[i], input_list[correct_pos] = input_list[correct_pos], input_list[i]
        swaps += 1
    else:
        i += 1

print(swaps)
```
