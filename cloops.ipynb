```python
# store the price (in dollars) tp the mobile in variab;e 'Price'
price = 900
for i in range(5,21,5):
   # calculate the discount amont
    discount = price * i/100
    print("The discount is", discount)
    # calculate the price after discount
    price_after_discount = price - discount
    print("The price after providing", i, "% discount is $", price_after_discount, sep = "")
    # print the discount price
    print("The price after discount is $", price_after_discount)
    print()
```

```python
# while loop
i = 5
while i <= 20:
    discount = price * i/100
    print("The discount is", discount)
    price_after_discount = price - discount
    print("The price after providing", i, "% discount is $", price_after_discount, sep = "")
    i += 5
    print("The price after discount is $", price_after_discount)
    print()
```
