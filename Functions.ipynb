```python
# write a functions that displays attribute of Applie Iphone

def display_details(name, color, price):
    '''Explains what the function does'''
    print(f"Name: {name}")
    print(f"Color: {color}")
    print(f"Price: {price}")

# call the function 
display_details("Iphone 13", "Space Gray", "$999")
```

```python
#lanmbda function
display_details = lambda name, color, price: print(f"Name: {name}\nColor: {color}\nPrice: {price}")
display_details("Iphone 13", "Space Gray", "$999")  
```
