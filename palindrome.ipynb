```python
pattern = input("Enter the pattern: ")
s = input("Enter the string: ")

words = s.split()
mapping = {}

if len(pattern) != len(words):
    print(False)
else:
    for p, w in zip(pattern, words):
        if p in mapping:
            if mapping[p] != w:
                print(False)
                break
        else:
            mapping[p] = w
    else:
        if len(set(mapping.values())) == len(mapping.values()):
            print(True)
        else:
            print(False)
```
