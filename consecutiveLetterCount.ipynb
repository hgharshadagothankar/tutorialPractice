```python
s = input()

result = ""
count = 1

for i in range(1, len(s)):
    if s[i] == s[i-1]:
        count += 1
    else:
        if count >= 3 and s[i-1] not in result:
            result += s[i-1]
        count = 1

# check last group
if count >= 3 and s[-1] not in result:
    result += s[-1]

print(result)
```
