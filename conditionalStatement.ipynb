```python
budget = int(input("Enter your budget: "))

if budget > 10000:  
    print("You can buy the Iphone")
else:
    print("Sorry! The mobile price is more than your budget")
    
```

```python
#loops

for i in range(10):
    print(i)
    
i = 0
while i < 10:
    print(i)
    i += 1  
    
```
