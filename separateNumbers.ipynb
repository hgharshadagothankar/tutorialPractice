```python
def separateNumbers(s):
    n = len(s)
    
    for i in range(1, n // 2 + 1):
        first = s[:i]
        
        if first[0] == '0':
            continue
        
        num = int(first)
        temp = first
        
        while len(temp) < n:
            num += 1
            temp += str(num)
        
        if temp == s:
            print("YES", first)
            return
    
    print("NO")


q = int(input())
for _ in range(q):
    s = input().strip()
    separateNumbers(s)
```
