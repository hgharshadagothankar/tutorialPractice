```python
#list
brand_list = ["Apple", "Google", "Microsoft"]
storage_list = [32, 64, 128]
price_list = [1000, 2000, 3000]

print(brand_list)
print(len(brand_list))
print(storage_list)
print(len(storage_list))
print(price_list)
print(len(price_list))
```

```python
#minimum price
print(min(price_list))
#maximum price
print(max(price_list))
#sum of price
print(sum(price_list))
print(price_list)
```
