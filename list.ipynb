```python
discounted_price_list = []
price_list = [100, 200, 300, 400, 500]


for price in price_list:
    discounted_price = price * (5/100)
    discounted_price_list.append(discounted_price)

print(discounted_price_list)
```

```python
#short hand using list comprehension
discounted_price_list = [price * (5/100) for price in price_list]
print(discounted_price_list)
```

```python
#customer budget
budget = 500
discounted_price_list = ["yes" if price < budget else "no" for price in price_list ]
print(discounted_price_list)
```
